```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(
    "../dataset/waterconsumption_noNaN_fixedTimechange (1).csv"
)

# Create a working copy
df_clean = df.copy()

print("Original dataset shape:", df_clean.shape)
```

```python
df_clean["Date"] = pd.to_datetime(
    df_clean["Date"],
    format="%m/%d/%Y"
)

print(df_clean["Date"].dtype)
print(df_clean["Date"].head())
```

```python
df_clean["Hour"] = df_clean["Time"].str[:2].astype(int)

display(df_clean[["Date", "Time", "Hour"]].head(10))
```

```python
# Extract the starting hour from the Time column
df_clean["Hour"] = (
    df_clean["Time"]
    .str.split("-").str[0]
    .str.split(":").str[0]
    .astype(int)
)

display(df_clean[["Date", "Time", "Hour"]].head(10))
```

```python
df_clean["Timestamp"] = (
    df_clean["Date"] +
    pd.to_timedelta(df_clean["Hour"], unit="h")
)

display(df_clean[["Date", "Time", "Hour", "Timestamp"]].head(10))
```

```python
# Extract date-based features

df_clean["Day"] = df_clean["Timestamp"].dt.day

df_clean["Month"] = df_clean["Timestamp"].dt.month

df_clean["Year"] = df_clean["Timestamp"].dt.year

df_clean["DayOfWeek"] = df_clean["Timestamp"].dt.dayofweek

df_clean["Weekday"] = df_clean["Timestamp"].dt.day_name()

# Display the result

display(
    df_clean[
        ["Timestamp", "Hour", "Day", "Month",
         "Year", "DayOfWeek", "Weekday"]
    ].head(10)
)
```

```python
print(df_clean[
    ["Timestamp", "Hour", "Day", "Month",
     "Year", "DayOfWeek", "Weekday"]
].head(5).to_string(index=False))
```

```python
# Check missing values before filling

print("Before filling:")
print(df_clean["Rain"].isnull().sum())

# Calculate median rainfall

rain_median = df_clean["Rain"].median()

print("Median Rainfall:", rain_median)

# Fill missing rainfall values

df_clean["Rain"] = df_clean["Rain"].fillna(rain_median)

# Check again

print("After filling:")
print(df_clean["Rain"].isnull().sum())
```

```python
print(df_clean.isnull().sum())
```

```python
# Check whether timestamps are in chronological order
print("Timestamps in correct order:",
      df_clean["Timestamp"].is_monotonic_increasing)

# Check duplicate timestamps
print("Duplicate timestamps:",
      df_clean["Timestamp"].duplicated().sum())

# Check the time difference between consecutive records
print("\nTime difference between records:")
print(df_clean["Timestamp"].diff().value_counts().head())
```

```python
print("Final dataset shape:", df_clean.shape)

print("\nColumn names:")
print(df_clean.columns.tolist())

print("\nMissing values:")
print(df_clean.isnull().sum())

print("\nFirst 5 records:")
display(df_clean.head())

print("\nLast 5 records:")
display(df_clean.tail())
```

```python
df_clean.to_csv(
    "../dataset/AquaSense_Cleaned_Data.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
